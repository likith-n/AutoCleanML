###  Classification Pipeline with AutoCleanML

This notebook showcases how **AutoCleanML** automates data preprocessing for classification tasks, allowing practitioners to focus on modeling rather than data cleaning.


```python
import pandas as pd 
from autocleanml import AutoCleanML
```

```python
# Download dataset from Kaggle
train = pd.read_csv('../data/titanic_train.csv')
test = pd.read_csv('../data/titanic_test.csv')

df = pd.concat([train, test], axis=0, ignore_index=True)
```

```python
df.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from sklearn.model_selection import cross_val_score
```

```python
clean = AutoCleanML(target='Embarked',model_type='linear')
```

```python
X_train,X_test,y_train,y_test,report = clean.fit_transform(df)
print(X_train.shape,X_test.shape)
print(y_train.shape,y_test.shape)
```

### Logistic Regression

```python
model1 = LogisticRegression()
model1.fit(X_train,y_train)
y_pred1 = model1.predict(X_test)
print("accuracy_score =",accuracy_score(y_test,y_pred1)*100)
```

### Decision Tree

```python
model2 = DecisionTreeClassifier(max_depth=8,random_state=42)
model2.fit(X_train,y_train)
y_pred2 = model2.predict(X_test)
print("Accuracy = ",accuracy_score(y_test,y_pred2)*100)
```

### Random Forest

```python
model3 = RandomForestClassifier(
    n_estimators=300,
    max_depth=10,
    random_state=42,
    n_jobs=-1
)

model3.fit(X_train,y_train)
y_pred3 = model3.predict(X_test)
print("Accuracy = ",accuracy_score(y_test,y_pred3)*100)
```

### Voting Ensemble

```python
from sklearn.ensemble import VotingClassifier

estimators=[('lr',model1),('dt',model2)]
model4 = VotingClassifier(
    estimators=estimators,
    n_jobs=-1
)

model4.fit(X_train,y_train)
y_pred4 = model4.predict(X_test)
print("accuracy_score",accuracy_score(y_test,y_pred4)*100)
```

### Boosting

```python
from sklearn.ensemble import GradientBoostingClassifier

model5 = GradientBoostingClassifier(
    n_estimators=500,
    max_depth=8,
    learning_rate=0.01,
    random_state=42
)
model5.fit(X_train,y_train)
y_pred5 = model5.predict(X_test)
print("Accuracy = ",accuracy_score(y_test,y_pred5)*100)
```
